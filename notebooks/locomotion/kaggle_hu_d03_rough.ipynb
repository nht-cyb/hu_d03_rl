# LimX HU_D03 rough terrain locomotion on Kaggle (mjlab), warm-started from the flat policy

Trains `Mjlab-Velocity-Rough-LimX-HU-D03`: the HU_D03 walks and runs on stairs, slopes, rough ground, waves and low
obstacles, with a human-like gait. It starts from the trained flat policy, `hu_d03_velocity_flat.pt`, instead of from
scratch.

### Before running

1. **Settings → Accelerator → GPU T4 x2.** Not P100: it is older than Turing and mjlab does not support it.
2. **Settings → Internet → On** (needs a phone-verified Kaggle account).

### How to run

* **Try it interactively first:** run the cells down to *5. Sanity checks*. The last video there is the warm-start
  policy on the new terrain, before any training: it should walk like the flat policy and stumble on obstacles.
* **Train with Save Version → Save & Run All (Commit).** That runs the whole notebook in the background (up to 12 h)
  and keeps everything in `/kaggle/working`. Training stops itself after `TIME_BUDGET_HOURS`.
* **See the results in section 8:** one overview video of every terrain, a measured results table, then each clip.

### Continue training in a new version

Open the notebook, *Add Input → Your Work → Notebooks*, and add **this notebook's previous version**. The settings
cell finds its newest checkpoint and resumes from it, with the terrain and speed curricula where they were.

### How the warm start works

A plain resume cannot load the flat checkpoint, for three reasons that `scripts/widen_flat_checkpoint.py` handles:

| Problem | What the script does |
|---|---|
| The rough task adds 187 height-scan inputs (actor 105 → 292, critic 117 → 304) | Adds them with zero weights, so the policy starts out acting exactly like the flat one and learns to use the scan |
| The observation normalizer has seen 1e9 samples and would barely adapt | Seeds the scan's statistics and lowers the sample count |
| The saved step counter (240,000) would start the speed curriculum at its last stage | Resets it, and the iteration count, to 0 |

It also clears Adam's state, whose shapes no longer fit, and raises the action noise to `WARM_START_STD`.

### What makes the gait human-like

There is no motion capture here: the gait comes from these rewards on top of the flat task's.

| Term | What it asks for |
|---|---|
| `contact_pattern` | Both feet down when standing, one foot when walking, at most one when running (allows a flight phase) |
| `air_time` | Long swings instead of short shuffling steps (it was off in the flat task) |
| `gait_symmetry` | Equal swing and stance times on both legs: no limp |
| `arm_leg_swing` | Each arm swings with the opposite leg |
| `leg_extension` | Walk tall, not crouched; measured from the lower foot, so it holds on slopes and stairs |
| `stance_width` | Feet about 22 cm apart |
| `mechanical_power` | Less wasted energy |
| `pose` | Tighter hips and shoulder roll (no flared arms), looser shoulder pitch and waist yaw |

The weights are a reasoned first guess, checked only for their relative size. If the result looks wrong, see
*Tuning the gait* at the end of the notebook.

Commands also change: mostly forward, little strafing, and up to 3 m/s (the flat task reached 2 m/s).

```python
import threading, time

def sh(cmd, cwd=None, max_seconds=None, check=True, log_file=None, on_line=None, on_tick=None):
  """Run a shell command and stream its output into the notebook.

  log_file: write the output there instead of the notebook (the last lines are
  shown if the command fails). on_line(line) sees every output line;
  on_tick(elapsed_s) is called every few seconds from the notebook thread.
  After max_seconds the command gets Ctrl-C (then a kill) and counts as done.
  """
  print("$", cmd, flush=True)
  log = open(log_file, "a") if log_file else None
  proc = subprocess.Popen(cmd, shell=True, cwd=cwd, text=True, bufsize=1,
                          stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                          start_new_session=True)
  sh.last_output = []

  def pump_output():
    for line in proc.stdout:
      if log:
        log.write(line)
      else:
        print(line, end="", flush=True)
      if on_line:
        on_line(line)
      sh.last_output.append(line)
      del sh.last_output[:-200]

  pump = threading.Thread(target=pump_output)
  pump.start()
  start, timed_out = time.time(), False
  while proc.poll() is None:
    if on_tick:
      on_tick(time.time() - start)
    if max_seconds is not None and time.time() - start > max_seconds:
      timed_out = True
      print(f"\n[time budget reached after {max_seconds / 3600:.1f} h: stopping]", flush=True)
      os.killpg(proc.pid, signal.SIGINT)
      try:
        proc.wait(timeout=120)
      except subprocess.TimeoutExpired:
        os.killpg(proc.pid, signal.SIGKILL)
        proc.wait()
      break
    time.sleep(10 if on_tick else 0.5)
  pump.join(timeout=30)
  if on_tick:
    on_tick(time.time() - start)
  if log:
    log.close()
  if check and proc.returncode != 0 and not timed_out:
    if log:
      print("".join(sh.last_output[-60:]))
    raise SystemExit(f"command failed with exit code {proc.returncode}")
  return proc.returncode

def fetch(repo, ref, dest):
  if not (dest / ".git").exists():
    sh(f"git init -q {dest} && git -C {dest} remote add origin {repo}")
  sh(f"git -C {dest} fetch -q --depth 1 origin {ref} && git -C {dest} checkout -q FETCH_HEAD")
  sh(f"git -C {dest} log -1 --format='%h %cd %s' --date=short")

fetch("https://github.com/mujocolab/mjlab.git", MJLAB_REF, CODE / "mjlab")
fetch("https://github.com/limxdynamics/humanoid-description.git", DESCRIPTION_REF, CODE / "humanoid-description")

sh("pip install -q uv")
sh(f"uv pip install --system -q -e '{CODE / 'mjlab'}[{CUDA_EXTRA}]'")
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/robot/hu_d03_spec.py
"""Load the LimX HU_D03 MJCF and turn it into a serial-joint model for mjlab.

This module only depends on `mujoco`, so it can be tested on a CPU-only
machine without mjlab installed.

What the conversion does:
  * Removes the parallel ankle (A/B achilles) and waist (A/B) linkages and the 6
    `<connect>` constraints that close them. The policy then commands the
    serial joints `ankle_pitch/roll` and `waist_roll/pitch` directly, like the
    Unitree G1 model in mjlab. The linkage mass is added to the parent body.
  * Removes the floor, light, camera, actuators and sensors from the source XML
    (mjlab adds its own terrain, actuators and sensors).
  * Sets every geom margin to 0 (MuJoCo Warp does not support margins on
    box/mesh pairs, such as the two foot boxes).
  * Names every collision geom (`<body>_collision<i>`, feet are
    `left_foot_collision` / `right_foot_collision`) so mjlab can match them.
  * Adds foot sites at the sole and the IMU sensors mjlab's tasks read.
"""

import os
from pathlib import Path

import mujoco
import numpy as np

_DEFAULT_DESCRIPTION_DIR = (
  Path(__file__).resolve().parents[3]
  / "third_party"
  / "humanoid-description"
  / "HU_D03_description"
)


def description_dir() -> Path:
  """Directory of `HU_D03_description` (override with $HU_D03_DESCRIPTION_DIR)."""
  return Path(os.environ.get("HU_D03_DESCRIPTION_DIR", _DEFAULT_DESCRIPTION_DIR))


def source_xml() -> Path:
  xml = description_dir() / "xml" / "HU_D03_03.xml"
  if not xml.exists():
    raise FileNotFoundError(
      f"{xml} not found. Clone https://github.com/limxdynamics/humanoid-description "
      "and set HU_D03_DESCRIPTION_DIR to its HU_D03_description folder."
    )
  return xml


# Bodies (with their subtrees) that form the closed-loop linkages.
LINKAGE_BODIES = (
  "left_A_achilles_link",
  "left_B_achilles_link",
  "right_A_achilles_link",
  "right_B_achilles_link",
  "waist_A_link",
  "waist_B_link",
)

# Foot sole: the foot box is centred 0.0535 m below the ankle roll frame and is
# 0.01 m half-thick, so the sole is at z = -0.0635.
FOOT_SITE_POS = (0.018, 0.0, -0.0635)
FOOT_GEOMS = {"left_foot": "left_foot_collision", "right_foot": "right_foot_collision"}


def _subtree_mass(body: mujoco.MjsBody) -> float:
  mass = body.mass
  for child in body.bodies:
    mass += _subtree_mass(child)
  return mass


def _add_mass(body: mujoco.MjsBody, extra: float) -> None:
  """Add point-like mass to a body, scaling its inertia by the same ratio."""
  ratio = (body.mass + extra) / body.mass
  body.mass = body.mass + extra
  body.inertia = np.asarray(body.inertia) * ratio


def get_spec() -> mujoco.MjSpec:
  spec = mujoco.MjSpec.from_file(str(source_xml()))

  # Equality constraints, actuators and sensors all reference linkage joints.
  for eq in list(spec.equalities):
    spec.delete(eq)
  for act in list(spec.actuators):
    spec.delete(act)
  for sensor in list(spec.sensors):
    spec.delete(sensor)

  # Floor, light and tracking camera live in the worldbody; mjlab adds terrain.
  for geom in list(spec.worldbody.geoms):
    spec.delete(geom)
  for light in list(spec.worldbody.lights):
    spec.delete(light)
  for cam in list(spec.worldbody.cameras):
    spec.delete(cam)

  # Remove the linkages, moving their mass to the parent body.
  for name in LINKAGE_BODIES:
    body = spec.body(name)
    parent = body.parent
    _add_mass(parent, _subtree_mass(body))
    spec.delete(body)

  # The free joint has a tiny spring/damper in the source XML; drop them.
  root = spec.body("base_link")
  free = root.joints[0]
  free.damping = np.zeros_like(free.damping)
  free.stiffness = np.zeros_like(free.stiffness)

  # Name collision geoms so mjlab's regex-based configs can find them.
  for body in spec.bodies:
    i = 0
    for geom in body.geoms:
      # The XML default gives every geom margin=0.001. MuJoCo Warp rejects a
      # non-zero margin on box/mesh pairs (e.g. foot-foot) with MULTICCD.
      geom.margin = 0.0
      if geom.contype == 0 and geom.conaffinity == 0:
        continue  # Visual mesh.
      if geom.name in FOOT_GEOMS:
        geom.name = FOOT_GEOMS[geom.name]
      elif not geom.name:
        geom.name = f"{body.name}_collision{i}"
        i += 1

  for side in ("left", "right"):
    spec.body(f"{side}_ankle_roll_link").add_site(
      name=f"{side}_foot", pos=FOOT_SITE_POS, size=(0.01, 0.01, 0.01), group=5
    )

  # Sensors read by mjlab's velocity and tracking tasks.
  spec.add_sensor(name="imu_ang_vel", type=mujoco.mjtSensor.mjSENS_GYRO,
                  objtype=mujoco.mjtObj.mjOBJ_SITE, objname="imu")
  spec.add_sensor(name="imu_lin_vel", type=mujoco.mjtSensor.mjSENS_VELOCIMETER,
                  objtype=mujoco.mjtObj.mjOBJ_SITE, objname="imu")
  spec.add_sensor(name="imu_lin_acc", type=mujoco.mjtSensor.mjSENS_ACCELEROMETER,
                  objtype=mujoco.mjtObj.mjOBJ_SITE, objname="imu")
  spec.add_sensor(name="imu_upvector", type=mujoco.mjtSensor.mjSENS_FRAMEZAXIS,
                  objtype=mujoco.mjtObj.mjOBJ_BODY, objname="world",
                  reftype=mujoco.mjtObj.mjOBJ_SITE, refname="imu")
  spec.add_sensor(name="root_angmom", type=mujoco.mjtSensor.mjSENS_SUBTREEANGMOM,
                  objtype=mujoco.mjtObj.mjOBJ_BODY, objname="base_link")

  return spec
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/robot/hu_d03_params.py
"""HU_D03 actuator and pose parameters (plain Python, no mjlab import).

Armatures are the reflected rotor inertias from the LimX SRDF
(`rotor mass * gear_ratio**2`), which match the `armature` values in the MJCF.
Effort limits are the `ctrlrange` of the MJCF motors.

LimX does not publish the PD gains used on the real robot, so stiffness and
damping follow mjlab's G1 rule: a second-order system with natural frequency
NATURAL_FREQ_HZ and damping ratio DAMPING_RATIO for the joint's armature.
Replace them with the real gains if you get them from LimX.

The ankle and waist roll/pitch joints are driven by two motors through a
parallel linkage. As for the G1, we assume a nominal 1:1 linkage, so these
joints get the armature and effort of both motors.
"""

import math
from dataclasses import dataclass

NATURAL_FREQ_HZ = 10.0
DAMPING_RATIO = 2.0

ARMATURE_LEG = 0.000244114 * 25.0**2  # 0.1526, hips and knees.
ARMATURE_36 = 0.000073217 * 36.0**2  # 0.0949, ankle/waist motors.
ARMATURE_ARM = 0.000073217 * 25.0**2  # 0.0458, shoulders and elbows.
ARMATURE_SMALL = 0.000017 * 25.0**2  # 0.0106, wrists, hands and head.


@dataclass(frozen=True)
class ActuatorGroup:
  joint_names_expr: tuple[str, ...]
  armature: float
  effort_limit: float

  @property
  def stiffness(self) -> float:
    w = 2.0 * math.pi * NATURAL_FREQ_HZ
    return self.armature * w**2

  @property
  def damping(self) -> float:
    w = 2.0 * math.pi * NATURAL_FREQ_HZ
    return 2.0 * DAMPING_RATIO * self.armature * w


ACTUATOR_GROUPS = {
  "leg": ActuatorGroup(
    (".*_hip_pitch_joint", ".*_hip_roll_joint", ".*_hip_yaw_joint", ".*_knee_joint"),
    ARMATURE_LEG,
    120.0,
  ),
  "ankle": ActuatorGroup(
    (".*_ankle_pitch_joint", ".*_ankle_roll_joint"), 2 * ARMATURE_36, 2 * 45.0
  ),
  "waist_yaw": ActuatorGroup(("waist_yaw_joint",), ARMATURE_36, 45.0),
  "waist": ActuatorGroup(
    ("waist_roll_joint", "waist_pitch_joint"), 2 * ARMATURE_36, 2 * 45.0
  ),
  "arm": ActuatorGroup(
    (
      ".*_shoulder_pitch_joint",
      ".*_shoulder_roll_joint",
      ".*_shoulder_yaw_joint",
      ".*_elbow_joint",
    ),
    ARMATURE_ARM,
    30.0,
  ),
  "small": ActuatorGroup(
    (".*_wrist_yaw_joint", ".*_wrist_pitch_joint", ".*_hand_yaw_joint", "head_.*"),
    ARMATURE_SMALL,
    18.0,
  ),
}

# Standing pose with slightly bent knees. Signs: negative hip pitch swings the
# thigh forward, positive knee flexes, negative elbow flexes. The base height
# puts the soles on the ground (checked by scripts/stand_test.py).
HOME_BASE_HEIGHT = 0.889
HOME_JOINT_POS = {
  ".*_hip_pitch_joint": -0.25,
  ".*_knee_joint": 0.5,
  ".*_ankle_pitch_joint": -0.25,
  "left_shoulder_roll_joint": 0.15,
  "right_shoulder_roll_joint": -0.15,
  ".*_elbow_joint": -0.3,
}
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/robot/hu_d03_constants.py
"""LimX HU_D03 robot config for mjlab."""

from mjlab.actuator import BuiltinPositionActuatorCfg
from mjlab.entity import EntityArticulationInfoCfg, EntityCfg
from mjlab.utils.spec_config import CollisionCfg

from hu_d03_mjlab.robot.hu_d03_params import (
  ACTUATOR_GROUPS,
  HOME_BASE_HEIGHT,
  HOME_JOINT_POS,
)
from hu_d03_mjlab.robot.hu_d03_spec import get_spec

##
# Actuator config.
##

HU_D03_ACTUATORS = tuple(
  BuiltinPositionActuatorCfg(
    target_names_expr=g.joint_names_expr,
    stiffness=g.stiffness,
    damping=g.damping,
    effort_limit=g.effort_limit,
    armature=g.armature,
  )
  for g in ACTUATOR_GROUPS.values()
)

##
# Keyframe config.
##

HOME_KEYFRAME = EntityCfg.InitialStateCfg(
  pos=(0, 0, HOME_BASE_HEIGHT),
  joint_pos=HOME_JOINT_POS,
  joint_vel={".*": 0.0},
)

##
# Collision config.
##

FOOT_GEOMS = r"^(left|right)_foot_collision$"

# All collisions, including self collisions. Feet get condim=3, the rest condim=1.
FULL_COLLISION = CollisionCfg(
  geom_names_expr=(".*_collision.*",),
  contype=1,
  conaffinity=1,
  condim={FOOT_GEOMS: 3, ".*_collision.*": 1},
  priority={FOOT_GEOMS: 1, ".*": 0},
  friction={FOOT_GEOMS: (0.6,)},
)

##
# Final config.
##

HU_D03_ARTICULATION = EntityArticulationInfoCfg(
  actuators=HU_D03_ACTUATORS,
  soft_joint_pos_limit_factor=0.9,
)


def get_hu_d03_robot_cfg() -> EntityCfg:
  """Get a fresh HU_D03 robot configuration instance."""
  return EntityCfg(
    init_state=HOME_KEYFRAME,
    collisions=(FULL_COLLISION,),
    spec_fn=get_spec,
    articulation=HU_D03_ARTICULATION,
  )


# With the 10 Hz gains, 0.25 * effort / stiffness (the G1 rule) gives only
# 0.03-0.05 rad per unit action on the legs, because the HU_D03's reflected
# inertias are large. Use a flat 0.25 rad instead.
HU_D03_ACTION_SCALE: dict[str, float] = {".*": 0.25}
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/tasks/gait_rewards.py
"""Reward terms that shape a human-like gait without a reference motion or phase clock."""

from __future__ import annotations

from typing import TYPE_CHECKING

import torch

from mjlab.entity import Entity
from mjlab.managers.scene_entity_config import SceneEntityCfg
from mjlab.sensor import ContactSensor
from mjlab.utils.lab_api.math import quat_apply_inverse

if TYPE_CHECKING:
  from mjlab.envs import ManagerBasedRlEnv


def _command_speed(env: ManagerBasedRlEnv, command_name: str) -> torch.Tensor:
  command = env.command_manager.get_command(command_name)
  assert command is not None
  return torch.norm(command[:, :2], dim=1) + torch.abs(command[:, 2])


def contact_pattern(
  env: ManagerBasedRlEnv,
  sensor_name: str,
  command_name: str,
  move_threshold: float = 0.1,
  run_threshold: float = 1.8,
) -> torch.Tensor:
  """Reward the foot contacts of the commanded gait.

  Standing: both feet down. Walking: one foot down, so the legs alternate; double
  support earns nothing but is not punished, and it stays brief. Running: at most
  one foot down, which also allows the flight phase.
  """
  sensor: ContactSensor = env.scene[sensor_name]
  found = sensor.data.found
  assert found is not None
  num_down = (found > 0).sum(dim=1)
  speed = _command_speed(env, command_name)
  return torch.where(
    speed < move_threshold,
    (num_down == 2).float(),
    torch.where(speed < run_threshold, num_down == 1, num_down <= 1).float(),
  )


def gait_symmetry(
  env: ManagerBasedRlEnv,
  sensor_name: str,
  command_name: str,
  command_threshold: float = 0.3,
) -> torch.Tensor:
  """Penalize a limp: both legs should have the same swing and stance times."""
  sensor: ContactSensor = env.scene[sensor_name]
  air, contact = sensor.data.last_air_time, sensor.data.last_contact_time
  assert air is not None and contact is not None
  cost = torch.square(air[:, 0] - air[:, 1]) + torch.square(contact[:, 0] - contact[:, 1])
  return cost * (_command_speed(env, command_name) > command_threshold).float()


def arm_leg_swing(
  env: ManagerBasedRlEnv,
  command_name: str,
  asset_cfg: SceneEntityCfg,
  gain: float = 0.7,
  std: float = 0.25,
  command_threshold: float = 0.3,
) -> torch.Tensor:
  """Reward swinging each arm with the opposite leg.

  asset_cfg.joint_names must be (left shoulder pitch, right shoulder pitch, left hip
  pitch, right hip pitch), with preserve_order=True. On the HU_D03 a negative angle
  swings both the arm and the thigh forward, so the signs match. Angles are measured
  from the default pose.
  """
  asset: Entity = env.scene[asset_cfg.name]
  q = asset.data.joint_pos[:, asset_cfg.joint_ids]
  q = q - asset.data.default_joint_pos[:, asset_cfg.joint_ids]
  error = torch.square(q[:, 0] - gain * q[:, 3]) + torch.square(q[:, 1] - gain * q[:, 2])
  moving = _command_speed(env, command_name) > command_threshold
  return torch.exp(-0.5 * error / std**2) * moving.float()


def stance_width(
  env: ManagerBasedRlEnv,
  target: float,
  asset_cfg: SceneEntityCfg,
  std: float = 0.05,
) -> torch.Tensor:
  """Reward feet at a natural lateral distance, measured in the pelvis frame."""
  asset: Entity = env.scene[asset_cfg.name]
  feet = asset.data.site_pos_w[:, asset_cfg.site_ids]
  between_w = feet[:, 0] - feet[:, 1]
  width = quat_apply_inverse(asset.data.root_link_quat_w, between_w)[:, 1]
  return torch.exp(-torch.square(width - target) / std**2)


def leg_extension(
  env: ManagerBasedRlEnv,
  target_height: float,
  command_name: str,
  asset_cfg: SceneEntityCfg,
  std: float = 0.04,
  command_threshold: float = 0.1,
) -> torch.Tensor:
  """Reward walking tall instead of crouching.

  Height is measured from the lower foot, not the world, so it means the same thing
  on a slope or on stairs.
  """
  asset: Entity = env.scene[asset_cfg.name]
  foot_z = asset.data.site_pos_w[:, asset_cfg.site_ids, 2].min(dim=1).values
  height = asset.data.root_link_pos_w[:, 2] - foot_z
  moving = _command_speed(env, command_name) > command_threshold
  return torch.exp(-torch.square(height - target_height) / std**2) * moving.float()


def mechanical_power(env: ManagerBasedRlEnv) -> torch.Tensor:
  """Penalize mechanical power: low-energy motion tends to look more natural."""
  asset: Entity = env.scene["robot"]
  return torch.sum(torch.abs(asset.data.qfrc_actuator * asset.data.joint_vel), dim=1)
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/tasks/rl_cfgs.py
"""PPO runner configs for the HU_D03 tasks (G1 hyperparameters)."""

from mjlab.rl import (
  RslRlModelCfg,
  RslRlOnPolicyRunnerCfg,
  RslRlPpoAlgorithmCfg,
)


def _runner_cfg(
  experiment_name: str, entropy_coef: float, save_interval: int
) -> RslRlOnPolicyRunnerCfg:
  return RslRlOnPolicyRunnerCfg(
    actor=RslRlModelCfg(
      hidden_dims=(512, 256, 128),
      activation="elu",
      obs_normalization=True,
      distribution_cfg={
        "class_name": "GaussianDistribution",
        "init_std": 1.0,
        "std_type": "scalar",
      },
    ),
    critic=RslRlModelCfg(
      hidden_dims=(512, 256, 128),
      activation="elu",
      obs_normalization=True,
    ),
    algorithm=RslRlPpoAlgorithmCfg(
      value_loss_coef=1.0,
      use_clipped_value_loss=True,
      clip_param=0.2,
      entropy_coef=entropy_coef,
      num_learning_epochs=5,
      num_mini_batches=4,
      learning_rate=1.0e-3,
      schedule="adaptive",
      gamma=0.99,
      lam=0.95,
      desired_kl=0.01,
      max_grad_norm=1.0,
    ),
    experiment_name=experiment_name,
    save_interval=save_interval,
    num_steps_per_env=24,
    max_iterations=30_000,
  )


def hu_d03_velocity_ppo_runner_cfg() -> RslRlOnPolicyRunnerCfg:
  return _runner_cfg("hu_d03_velocity", entropy_coef=0.01, save_interval=50)


def hu_d03_velocity_rough_ppo_runner_cfg() -> RslRlOnPolicyRunnerCfg:
  return _runner_cfg("hu_d03_velocity_rough", entropy_coef=0.01, save_interval=50)


def hu_d03_tracking_ppo_runner_cfg() -> RslRlOnPolicyRunnerCfg:
  return _runner_cfg("hu_d03_tracking", entropy_coef=0.005, save_interval=100)
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/tasks/velocity_env_cfgs.py
"""LimX HU_D03 velocity environment configurations (adapted from mjlab's G1 ones)."""

from mjlab.envs import ManagerBasedRlEnvCfg
from mjlab.envs import mdp as envs_mdp
from mjlab.envs.mdp.actions import JointPositionActionCfg
from mjlab.managers.event_manager import EventTermCfg
from mjlab.managers.reward_manager import RewardTermCfg
from mjlab.managers.scene_entity_config import SceneEntityCfg
from mjlab.sensor import (
  ContactMatch,
  ContactSensorCfg,
  ObjRef,
  RayCastSensorCfg,
  RingPatternCfg,
  TerrainHeightSensorCfg,
)
from mjlab.tasks.velocity import mdp
from mjlab.tasks.velocity.mdp import UniformVelocityCommandCfg
from mjlab.tasks.velocity.velocity_env_cfg import make_velocity_env_cfg
from mjlab.terrains import config as terrain_cfg

from hu_d03_mjlab.robot.hu_d03_constants import (
  FOOT_GEOMS,
  HU_D03_ACTION_SCALE,
  get_hu_d03_robot_cfg,
)
from hu_d03_mjlab.tasks import gait_rewards

PELVIS = "base_link"
TORSO = "waist_pitch_link"


def _hu_d03_base_env_cfg(play: bool = False) -> ManagerBasedRlEnvCfg:
  """Settings shared by the flat and rough tasks."""
  cfg = make_velocity_env_cfg()

  # mjlab's G1 uses 500 iterations and 70 contacts per env, too much for a T4's 15 GB. The
  # collision buffers need about 40 kB (at 300 iterations) times nconmax * num_envs, once
  # for each of the step and forward CUDA graphs: about 10 GB here, on top of about 2 GB.
  cfg.sim.mujoco.ccd_iterations = 300
  cfg.sim.contact_sensor_maxmatch = 500
  # A pool shared by all envs, not a per-env cap; envs use about 4 on average.
  cfg.sim.nconmax = 30

  cfg.scene.entities = {"robot": get_hu_d03_robot_cfg()}

  for sensor in cfg.scene.sensors or ():
    if sensor.name == "terrain_scan":
      assert isinstance(sensor, RayCastSensorCfg)
      assert isinstance(sensor.frame, ObjRef)
      sensor.frame.name = PELVIS

  site_names = ("left_foot", "right_foot")

  for sensor in cfg.scene.sensors or ():
    if sensor.name == "foot_height_scan":
      assert isinstance(sensor, TerrainHeightSensorCfg)
      sensor.frame = tuple(
        ObjRef(type="site", name=s, entity="robot") for s in site_names
      )
      sensor.pattern = RingPatternCfg.single_ring(radius=0.03, num_samples=6)

  feet_ground_cfg = ContactSensorCfg(
    name="feet_ground_contact",
    primary=ContactMatch(
      mode="subtree",
      pattern=r"^(left_ankle_roll_link|right_ankle_roll_link)$",
      entity="robot",
    ),
    secondary=ContactMatch(mode="body", pattern="terrain"),
    fields=("found", "force"),
    reduce="netforce",
    num_slots=1,
    track_air_time=True,
  )
  self_collision_cfg = ContactSensorCfg(
    name="self_collision",
    primary=ContactMatch(mode="subtree", pattern=PELVIS, entity="robot"),
    secondary=ContactMatch(mode="subtree", pattern=PELVIS, entity="robot"),
    fields=("found", "force"),
    reduce="none",
    num_slots=1,
    history_length=4,
  )
  cfg.scene.sensors = (cfg.scene.sensors or ()) + (
    feet_ground_cfg,
    self_collision_cfg,
  )

  if cfg.scene.terrain is not None and cfg.scene.terrain.terrain_generator is not None:
    cfg.scene.terrain.terrain_generator.curriculum = True

  joint_pos_action = cfg.actions["joint_pos"]
  assert isinstance(joint_pos_action, JointPositionActionCfg)
  joint_pos_action.scale = HU_D03_ACTION_SCALE

  cfg.viewer.body_name = TORSO

  twist_cmd = cfg.commands["twist"]
  assert isinstance(twist_cmd, UniformVelocityCommandCfg)
  twist_cmd.viz.z_offset = 1.75

  cfg.events["foot_friction"].params["asset_cfg"].geom_names = FOOT_GEOMS
  cfg.events["base_com"].params["asset_cfg"].body_names = (TORSO,)

  # Same std rationale as the G1: loose knees/hip pitch, tight ankle roll and
  # waist roll/pitch, moderate arms. The HU_D03 also has a head and hand yaw.
  cfg.rewards["pose"].params["std_standing"] = {".*": 0.05}
  cfg.rewards["pose"].params["std_walking"] = {
    # Lower body.
    r".*hip_pitch.*": 0.3,
    r".*hip_roll.*": 0.15,
    r".*hip_yaw.*": 0.15,
    r".*knee.*": 0.35,
    r".*ankle_pitch.*": 0.25,
    r".*ankle_roll.*": 0.1,
    # Waist and head.
    r".*waist_yaw.*": 0.2,
    r".*waist_roll.*": 0.08,
    r".*waist_pitch.*": 0.1,
    r"head_.*": 0.1,
    # Arms.
    r".*shoulder_pitch.*": 0.15,
    r".*shoulder_roll.*": 0.15,
    r".*shoulder_yaw.*": 0.1,
    r".*elbow.*": 0.15,
    r".*wrist.*": 0.3,
    r".*hand_yaw.*": 0.3,
  }
  cfg.rewards["pose"].params["std_running"] = {
    # Lower body.
    r".*hip_pitch.*": 0.5,
    r".*hip_roll.*": 0.2,
    r".*hip_yaw.*": 0.2,
    r".*knee.*": 0.6,
    r".*ankle_pitch.*": 0.35,
    r".*ankle_roll.*": 0.15,
    # Waist and head.
    r".*waist_yaw.*": 0.3,
    r".*waist_roll.*": 0.08,
    r".*waist_pitch.*": 0.2,
    r"head_.*": 0.1,
    # Arms.
    r".*shoulder_pitch.*": 0.5,
    r".*shoulder_roll.*": 0.2,
    r".*shoulder_yaw.*": 0.15,
    r".*elbow.*": 0.35,
    r".*wrist.*": 0.3,
    r".*hand_yaw.*": 0.3,
  }

  cfg.rewards["upright"].params["asset_cfg"].body_names = (TORSO,)
  cfg.rewards["body_ang_vel"].params["asset_cfg"].body_names = (TORSO,)

  for reward_name in ["foot_clearance", "foot_slip"]:
    cfg.rewards[reward_name].params["asset_cfg"].site_names = site_names

  cfg.rewards["body_ang_vel"].weight = -0.05
  cfg.rewards["angular_momentum"].weight = -0.02
  cfg.rewards["air_time"].weight = 0.0

  cfg.rewards["self_collisions"] = RewardTermCfg(
    func=mdp.self_collision_cost,
    weight=-1.0,
    params={"sensor_name": self_collision_cfg.name, "force_threshold": 10.0},
  )

  if play:
    cfg.episode_length_s = int(1e9)

    cfg.observations["actor"].enable_corruption = False
    cfg.events.pop("push_robot", None)
    cfg.terminations.pop("out_of_terrain_bounds", None)
    cfg.curriculum = {}
    cfg.events["randomize_terrain"] = EventTermCfg(
      func=envs_mdp.randomize_terrain,
      mode="reset",
      params={},
    )

    if cfg.scene.terrain is not None:
      if cfg.scene.terrain.terrain_generator is not None:
        cfg.scene.terrain.terrain_generator.curriculum = False
        cfg.scene.terrain.terrain_generator.num_cols = 5
        cfg.scene.terrain.terrain_generator.num_rows = 5
        cfg.scene.terrain.terrain_generator.border_width = 10.0

  return cfg


def hu_d03_rough_env_cfg(play: bool = False) -> ManagerBasedRlEnvCfg:
  """Create LimX HU_D03 rough terrain velocity configuration with a human-like gait.

  Meant to be warm-started from a flat policy (scripts/widen_flat_checkpoint.py), so
  it starts on easy terrain and at speeds that policy already handles.
  """
  cfg = _hu_d03_base_env_cfg(play=play)

  assert cfg.scene.terrain is not None
  terrain = cfg.scene.terrain.terrain_generator
  assert terrain is not None
  # Terrains a person walks on; mjlab's 45 degree slopes would force an unnatural gait.
  terrain.sub_terrains = {
    "flat": terrain_cfg.flat(proportion=0.15),
    # Robots spawn at the center of a tile: a pyramid's top, or the bottom of an inverted one.
    "stairs_down": terrain_cfg.pyramid_stairs(
      proportion=0.2, step_height_range=(0.02, 0.15), step_width=0.35
    ),
    "stairs_up": terrain_cfg.pyramid_stairs_inv(
      proportion=0.2, step_height_range=(0.02, 0.15), step_width=0.35
    ),
    "slope_down": terrain_cfg.hf_pyramid_slope(proportion=0.1, slope_range=(0.0, 0.45)),
    "slope_up": terrain_cfg.hf_pyramid_slope_inv(
      proportion=0.1, slope_range=(0.0, 0.45)
    ),
    "random_rough": terrain_cfg.random_rough(proportion=0.1, noise_range=(0.01, 0.06)),
    "waves": terrain_cfg.wave_terrain(proportion=0.1, amplitude_range=(0.0, 0.12)),
    "obstacles": terrain_cfg.discrete_obstacles(
      proportion=0.05, obstacle_height_range=(0.02, 0.12), num_obstacles=20
    ),
  }
  cfg.scene.terrain.max_init_terrain_level = 2

  # People walk forward far more than they strafe or back up, and they also run.
  twist_cmd = cfg.commands["twist"]
  assert isinstance(twist_cmd, UniformVelocityCommandCfg)
  twist_cmd.rel_forward_envs = 0.4
  twist_cmd.ranges.lin_vel_x = (-0.8, 1.5)
  twist_cmd.ranges.lin_vel_y = (-0.4, 0.4)
  if not play:
    cfg.curriculum["command_vel"].params["velocity_stages"] = [
      {"step": 0, "lin_vel_x": (-0.8, 1.5), "ang_vel_z": (-0.5, 0.5)},
      {"step": 1500 * 24, "lin_vel_x": (-1.0, 2.0), "ang_vel_z": (-0.7, 0.7)},
      {"step": 4000 * 24, "lin_vel_x": (-1.0, 3.0), "lin_vel_y": (-0.5, 0.5)},
    ]

  # Tighter on the joints that made the flat gait stiff-armed and wide-hipped, looser
  # on the shoulder pitch and waist yaw so that the arms and torso can counter-swing.
  pose = cfg.rewards["pose"].params
  pose["std_walking"] |= {
    r".*hip_roll.*": 0.12,
    r".*hip_yaw.*": 0.1,
    r".*shoulder_pitch.*": 0.4,
    r".*shoulder_roll.*": 0.08,
    r".*shoulder_yaw.*": 0.08,
    r".*elbow.*": 0.2,
    r"head_.*": 0.05,
    r".*waist_yaw.*": 0.25,
  }
  pose["std_running"] |= {
    r".*hip_yaw.*": 0.15,
    r".*shoulder_pitch.*": 0.7,
    r".*shoulder_roll.*": 0.12,
    r".*shoulder_yaw.*": 0.12,
    r".*elbow.*": 0.5,
    r"head_.*": 0.05,
  }
  pose["running_threshold"] = 1.8

  cfg.rewards["upright"].weight = 1.5
  cfg.rewards["air_time"].weight = 0.25
  cfg.rewards["air_time"].params["threshold_min"] = 0.1
  cfg.rewards["air_time"].params["command_threshold"] = 0.3
  cfg.rewards["contact_pattern"] = RewardTermCfg(
    func=gait_rewards.contact_pattern,
    weight=0.5,
    params={"sensor_name": "feet_ground_contact", "command_name": "twist"},
  )
  cfg.rewards["gait_symmetry"] = RewardTermCfg(
    func=gait_rewards.gait_symmetry,
    weight=-2.0,
    params={"sensor_name": "feet_ground_contact", "command_name": "twist"},
  )
  cfg.rewards["arm_leg_swing"] = RewardTermCfg(
    func=gait_rewards.arm_leg_swing,
    weight=0.75,
    params={
      "command_name": "twist",
      "std": 0.2,
      "asset_cfg": SceneEntityCfg(
        "robot",
        joint_names=(
          "left_shoulder_pitch_joint",
          "right_shoulder_pitch_joint",
          "left_hip_pitch_joint",
          "right_hip_pitch_joint",
        ),
        preserve_order=True,
      ),
    },
  )
  feet = SceneEntityCfg("robot", site_names=("left_foot", "right_foot"), preserve_order=True)
  cfg.rewards["stance_width"] = RewardTermCfg(
    func=gait_rewards.stance_width,
    weight=0.25,
    params={"target": 0.22, "asset_cfg": feet},
  )
  # Home pose: 0.889 m with 0.5 rad knees; straight legs: 0.912 m.
  cfg.rewards["leg_extension"] = RewardTermCfg(
    func=gait_rewards.leg_extension,
    weight=0.5,
    params={"target_height": 0.9, "command_name": "twist", "asset_cfg": feet},
  )
  cfg.rewards["mechanical_power"] = RewardTermCfg(
    func=gait_rewards.mechanical_power,
    weight=-5.0e-4,
  )

  return cfg


def hu_d03_flat_env_cfg(play: bool = False) -> ManagerBasedRlEnvCfg:
  """Create LimX HU_D03 flat terrain velocity configuration."""
  cfg = _hu_d03_base_env_cfg(play=play)

  cfg.sim.njmax = 300
  cfg.sim.mujoco.ccd_iterations = 50
  cfg.sim.contact_sensor_maxmatch = 64
  cfg.sim.nconmax = None

  assert cfg.scene.terrain is not None
  cfg.scene.terrain.terrain_type = "plane"
  cfg.scene.terrain.terrain_generator = None

  cfg.scene.sensors = tuple(
    s for s in (cfg.scene.sensors or ()) if s.name != "terrain_scan"
  )
  del cfg.observations["actor"].terms["height_scan"]
  del cfg.observations["critic"].terms["height_scan"]

  cfg.terminations.pop("out_of_terrain_bounds", None)
  cfg.curriculum.pop("terrain_levels", None)

  if play:
    twist_cmd = cfg.commands["twist"]
    assert isinstance(twist_cmd, UniformVelocityCommandCfg)
    twist_cmd.ranges.lin_vel_x = (-1.5, 2.0)
    twist_cmd.ranges.ang_vel_z = (-0.7, 0.7)

  return cfg
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/src/hu_d03_mjlab/tasks/tracking_env_cfgs.py
"""LimX HU_D03 flat tracking (mimic) environment configurations."""

from mjlab.envs import ManagerBasedRlEnvCfg
from mjlab.envs.mdp.actions import JointPositionActionCfg
from mjlab.managers.observation_manager import ObservationGroupCfg
from mjlab.sensor import ContactMatch, ContactSensorCfg
from mjlab.tasks.tracking.mdp import MotionCommandCfg
from mjlab.tasks.tracking.tracking_env_cfg import make_tracking_env_cfg

from hu_d03_mjlab.robot.hu_d03_constants import (
  FOOT_GEOMS,
  HU_D03_ACTION_SCALE,
  get_hu_d03_robot_cfg,
)

PELVIS = "base_link"
TORSO = "waist_pitch_link"

# Bodies whose pose is tracked (same choice of links as the G1 config).
TRACKED_BODIES = (
  PELVIS,
  "left_hip_roll_link",
  "left_knee_link",
  "left_ankle_roll_link",
  "right_hip_roll_link",
  "right_knee_link",
  "right_ankle_roll_link",
  TORSO,
  "left_shoulder_roll_link",
  "left_elbow_link",
  "left_wrist_pitch_link",
  "right_shoulder_roll_link",
  "right_elbow_link",
  "right_wrist_pitch_link",
)


def hu_d03_flat_tracking_env_cfg(
  has_state_estimation: bool = True,
  play: bool = False,
) -> ManagerBasedRlEnvCfg:
  """Create LimX HU_D03 flat terrain tracking configuration."""
  cfg = make_tracking_env_cfg()

  cfg.scene.entities = {"robot": get_hu_d03_robot_cfg()}

  self_collision_cfg = ContactSensorCfg(
    name="self_collision",
    primary=ContactMatch(mode="subtree", pattern=PELVIS, entity="robot"),
    secondary=ContactMatch(mode="subtree", pattern=PELVIS, entity="robot"),
    fields=("found", "force"),
    reduce="none",
    num_slots=1,
    history_length=4,
  )
  cfg.scene.sensors = (self_collision_cfg,)

  joint_pos_action = cfg.actions["joint_pos"]
  assert isinstance(joint_pos_action, JointPositionActionCfg)
  joint_pos_action.scale = HU_D03_ACTION_SCALE

  motion_cmd = cfg.commands["motion"]
  assert isinstance(motion_cmd, MotionCommandCfg)
  motion_cmd.anchor_body_name = TORSO
  motion_cmd.body_names = TRACKED_BODIES

  cfg.events["foot_friction"].params["asset_cfg"].geom_names = FOOT_GEOMS
  cfg.events["base_com"].params["asset_cfg"].body_names = (TORSO,)

  cfg.terminations["ee_body_pos"].params["body_names"] = (
    "left_ankle_roll_link",
    "right_ankle_roll_link",
    "left_wrist_pitch_link",
    "right_wrist_pitch_link",
  )

  cfg.viewer.body_name = TORSO

  if not has_state_estimation:
    new_actor_terms = {
      k: v
      for k, v in cfg.observations["actor"].terms.items()
      if k not in ["motion_anchor_pos_b", "base_lin_vel"]
    }
    cfg.observations["actor"] = ObservationGroupCfg(
      terms=new_actor_terms,
      concatenate_terms=True,
      enable_corruption=True,
    )

  if play:
    cfg.episode_length_s = int(1e9)

    cfg.observations["actor"].enable_corruption = False
    cfg.events.pop("push_robot", None)

    motion_cmd.pose_range = {}
    motion_cmd.velocity_range = {}

    motion_cmd.sampling_mode = "start"

  return cfg
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/scripts/stand_test.py
"""CPU stand test for the converted HU_D03 model (plain MuJoCo, no GPU, no mjlab).

Adds PD position actuators with the gains from hu_d03_params, puts the robot in
the home pose with its soles on the ground, and holds the pose for a few
seconds. Prints the home base height, any self-contacts at the home pose, and
how far the base drifts.

  python scripts/stand_test.py [--seconds 5] [--viewer]
"""

import argparse
import re
import sys
from pathlib import Path

import mujoco
import numpy as np

sys.path.insert(0, str(Path(__file__).resolve().parents[1] / "src"))

from hu_d03_mjlab.robot.hu_d03_params import ACTUATOR_GROUPS, HOME_JOINT_POS  # noqa: E402
from hu_d03_mjlab.robot.hu_d03_spec import get_spec  # noqa: E402


def match(patterns, name):
  return any(re.fullmatch(p, name) for p in patterns)


def build_model():
  spec = get_spec()
  spec.worldbody.add_geom(
    name="floor", type=mujoco.mjtGeom.mjGEOM_PLANE, size=(0, 0, 1),
  )
  spec.option.timestep = 0.005
  for joint in spec.joints:
    if joint.type != mujoco.mjtJoint.mjJNT_HINGE:
      continue
    group = next(g for g in ACTUATOR_GROUPS.values() if match(g.joint_names_expr, joint.name))
    joint.armature = group.armature
    act = spec.add_actuator(name=joint.name, target=joint.name,
                            trntype=mujoco.mjtTrn.mjTRN_JOINT)
    act.set_to_position(kp=group.stiffness, kv=group.damping)
    act.forcelimited = True
    act.forcerange = (-group.effort_limit, group.effort_limit)
  return spec.compile()


def home_qpos(model):
  qpos = model.qpos0.copy()
  for j in range(1, model.njnt):
    name = model.joint(j).name
    for pattern, value in HOME_JOINT_POS.items():
      if re.fullmatch(pattern, name):
        qpos[model.jnt_qposadr[j]] = value
  # Lift or lower the base so the lowest foot site touches z = 0.
  data = mujoco.MjData(model)
  data.qpos[:] = qpos
  data.qpos[2] = 1.0
  mujoco.mj_kinematics(model, data)
  sole = min(data.site("left_foot").xpos[2], data.site("right_foot").xpos[2])
  qpos[2] = 1.0 - sole
  return qpos


def main():
  parser = argparse.ArgumentParser()
  parser.add_argument("--seconds", type=float, default=5.0)
  parser.add_argument("--viewer", action="store_true")
  args = parser.parse_args()

  model = build_model()
  data = mujoco.MjData(model)
  qpos = home_qpos(model)
  data.qpos[:] = qpos
  data.ctrl[:] = qpos[7:]
  mujoco.mj_forward(model, data)

  print(f"home base height: {qpos[2]:.4f} m, total mass {model.body_mass.sum():.2f} kg")
  floor = model.geom("floor").id
  self_contacts = [
    (model.geom(c.geom1).name, model.geom(c.geom2).name)
    for c in data.contact[: data.ncon]
    if floor not in (c.geom1, c.geom2)
  ]
  print(f"self-contacts at home pose: {self_contacts or 'none'}")

  def step():
    mujoco.mj_step(model, data)

  if args.viewer:
    from mujoco import viewer as mj_viewer

    with mj_viewer.launch_passive(model, data) as v:
      while v.is_running() and data.time < args.seconds:
        step()
        v.sync()
  else:
    while data.time < args.seconds:
      step()

  drift = np.linalg.norm(data.qpos[:2] - qpos[:2])
  up = data.xmat[model.body("base_link").id].reshape(3, 3)[:, 2]
  tilt = np.degrees(np.arccos(np.clip(up[2], -1, 1)))
  joint_err = np.abs(data.qpos[7:] - qpos[7:]).max()
  print(f"after {args.seconds:.1f}s: base z {data.qpos[2]:.4f} m, xy drift {drift:.4f} m, "
        f"tilt {tilt:.2f} deg, max joint error {joint_err:.4f} rad")
  ok = data.qpos[2] > qpos[2] - 0.1 and tilt < 10
  print("STAND TEST", "PASSED" if ok else "FAILED")
  return 0 if ok else 1


if __name__ == "__main__":
  sys.exit(main())
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/scripts/record_video.py
"""Roll out a policy headless and save an .mp4 (for Colab/Kaggle, where mjlab's
`play` viewer cannot open a window).

  python scripts/record_video.py Mjlab-Velocity-Flat-LimX-HU-D03 \
      --checkpoint-file logs/.../model_5000.pt --output-file walk.mp4
  python scripts/record_video.py Mjlab-Tracking-Flat-LimX-HU-D03 \
      --checkpoint-file ... --motion-file motion.npz --output-file mimic.mp4
  # --agent zero holds the default pose (stand test, no checkpoint needed).
  # --start-seconds 42 starts a tracking video at that time of the motion (e.g. one dance of a medley).
  # --forward-speed 2.5 holds a straight-line command; --terrain stairs_up --difficulty 1 keeps one Rough
  # terrain at one difficulty. --num-envs 64 --stats-file out.json also measures falls and speed over
  # all those robots; the camera follows robot 0.
  # Defaults to 1280x720; e.g. --width 1920 --height 1080 --distance 4 for 1080p.
"""

import json
from dataclasses import asdict
from pathlib import Path
from typing import Literal

import mediapy
import torch
import tyro

import mjlab
from mjlab.envs import ManagerBasedRlEnv
from mjlab.rl import MjlabOnPolicyRunner, RslRlVecEnvWrapper
from mjlab.tasks.registry import load_env_cfg, load_rl_cfg, load_runner_cls
from mjlab.tasks.tracking.mdp import MotionCommandCfg


def main(
  task_id: tyro.conf.Positional[str],
  output_file: str,
  checkpoint_file: str | None = None,
  motion_file: str | None = None,
  agent: Literal["trained", "zero"] = "trained",
  seconds: float = 10.0,
  start_seconds: float = 0.0,
  no_terminations: bool = False,
  width: int = 1280,
  height: int = 720,
  distance: float = 3.0,
  elevation: float = -15.0,
  azimuth: float = 90.0,
  forward_speed: float | None = None,
  terrain: str | None = None,
  difficulty: float | None = None,
  num_envs: int = 1,
  stats_file: str | None = None,
):
  device = "cuda:0" if torch.cuda.is_available() else "cpu"
  env_cfg = load_env_cfg(task_id, play=True)
  agent_cfg = load_rl_cfg(task_id)
  env_cfg.scene.num_envs = num_envs
  env_cfg.viewer.max_extra_envs = 0
  if forward_speed is not None:
    twist = env_cfg.commands["twist"]
    twist.rel_standing_envs = twist.rel_heading_envs = twist.rel_forward_envs = 0.0
    twist.ranges.lin_vel_x = (forward_speed, forward_speed)
    twist.ranges.lin_vel_y = twist.ranges.ang_vel_z = (0.0, 0.0)
    env_cfg.curriculum = {}
  if terrain is not None:
    generator = env_cfg.scene.terrain.terrain_generator
    generator.sub_terrains = {terrain: generator.sub_terrains[terrain]}
    if difficulty is not None:
      generator.difficulty_range = (difficulty, difficulty)
  # mjlab renders 320x240 from 5 m away by default; frame the whole robot instead.
  env_cfg.viewer.width, env_cfg.viewer.height = width, height
  env_cfg.viewer.distance = distance
  env_cfg.viewer.elevation = elevation
  env_cfg.viewer.azimuth = azimuth
  if no_terminations:
    env_cfg.terminations = {}

  motion_cmd = env_cfg.commands.get("motion")
  if isinstance(motion_cmd, MotionCommandCfg):
    if motion_file is None:
      raise ValueError("Tracking tasks need --motion-file.")
    motion_cmd.motion_file = motion_file

  env = ManagerBasedRlEnv(cfg=env_cfg, device=device, render_mode="rgb_array")
  env = RslRlVecEnvWrapper(env, clip_actions=agent_cfg.clip_actions)

  if agent == "zero":
    shape = env.unwrapped.action_space.shape

    def policy(obs):
      del obs
      return torch.zeros(shape, device=env.unwrapped.device)
  else:
    if checkpoint_file is None:
      raise ValueError("--checkpoint-file is required with --agent trained.")
    runner_cls = load_runner_cls(task_id) or MjlabOnPolicyRunner
    runner = runner_cls(env, asdict(agent_cfg), device=device)
    runner.load(checkpoint_file, load_cfg={"actor": True}, strict=True,
                map_location=device)
    policy = runner.get_inference_policy(device=device)

  if start_seconds > 0:
    # Put the robot exactly in the reference state at that frame (play mode starts at frame 0).
    command = env.unwrapped.command_manager.get_term("motion")
    frame = round(start_seconds / env.unwrapped.step_dt)
    command.reset_to_frame(torch.arange(env.num_envs, device=env.unwrapped.device), frame)
    env.unwrapped.sim.forward()
    command.update_relative_body_poses()

  obs = env.get_observations()
  frames = []
  steps = int(seconds / env.unwrapped.step_dt)
  # Play mode has no time-out, so every reset is a fall.
  falls = torch.zeros(env.num_envs, device=env.unwrapped.device)
  speed = 0.0
  with torch.inference_mode():
    for _ in range(steps):
      obs, _, dones, _ = env.step(policy(obs))
      frames.append(env.unwrapped.render())
      falls += dones.float()
      if stats_file:
        forward = env.unwrapped.scene["robot"].data.root_link_lin_vel_b[:, 0]
        speed += forward.mean().item() / steps

  fps = round(1.0 / env.unwrapped.step_dt)
  Path(output_file).parent.mkdir(parents=True, exist_ok=True)
  mediapy.write_video(output_file, frames, fps=fps)
  print(f"wrote {output_file} ({len(frames)} frames at {fps} fps)")
  if stats_file:
    stats = {
      "num_envs": env.num_envs,
      "seconds": seconds,
      "fell_percent": 100.0 * (falls > 0).float().mean().item(),
      "camera_robot_fell": bool(falls[0] > 0),
      "mean_forward_speed": speed,
    }
    Path(stats_file).parent.mkdir(parents=True, exist_ok=True)
    Path(stats_file).write_text(json.dumps(stats))
    print(stats)
  env.close()


if __name__ == "__main__":
  tyro.cli(main, config=mjlab.TYRO_FLAGS)
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/scripts/widen_flat_checkpoint.py
"""Turn a Flat velocity checkpoint into a warm start for the Rough task.

The Rough observations add a terrain height scan, so the first layer of the actor
and critic is wider and a plain `--agent.resume` cannot load a Flat checkpoint.
This inserts the new inputs with zero weights: the widened policy walks exactly like
the Flat one until training teaches it to use the scan.

  python scripts/widen_flat_checkpoint.py results/policies/hu_d03_velocity_flat.pt \
      logs/rsl_rl/hu_d03_velocity_rough/0000_warm_start_flat/model_0.pt
"""

from pathlib import Path

import torch
import tyro

FLAT_ACTOR_OBS = 105
FLAT_CRITIC_OBS = 117
# Both groups list height_scan straight after the 105 shared terms; the critic's
# foot terms follow it. 17 x 11 rays: GridPatternCfg(size=(1.6, 1.0), resolution=0.1).
SCAN_START = 105
SCAN_RAYS = 187
# The scan is (base height above terrain) / 5 m, see make_velocity_env_cfg. The
# normalizer's real statistics take over during training; these keep the first
# iterations sensible (measured on the Rough terrains with the Flat policy).
SCAN_MEAN = 0.17
SCAN_STD = 0.03
# The Flat run's normalizer has seen 1e9 samples and would barely move on new terrain.
# At 4096 envs this adapts it over roughly the first 500 iterations.
NORMALIZER_COUNT = 50_000_000


def _insert(tensor: torch.Tensor, dim: int, value: float) -> torch.Tensor:
  head, tail = tensor.split([SCAN_START, tensor.shape[dim] - SCAN_START], dim)
  shape = list(tensor.shape)
  shape[dim] = SCAN_RAYS
  return torch.cat([head, tensor.new_full(shape, value), tail], dim)


def _widen(state: dict, flat_obs: int) -> None:
  width = state["mlp.0.weight"].shape[1]
  if width != flat_obs:
    raise SystemExit(f"Not a Flat checkpoint: {width} inputs instead of {flat_obs}.")
  state["mlp.0.weight"] = _insert(state["mlp.0.weight"], 1, 0.0)
  state["obs_normalizer._mean"] = _insert(state["obs_normalizer._mean"], 1, SCAN_MEAN)
  state["obs_normalizer._std"] = _insert(state["obs_normalizer._std"], 1, SCAN_STD)
  state["obs_normalizer._var"] = _insert(state["obs_normalizer._var"], 1, SCAN_STD**2)
  state["obs_normalizer.count"] = state["obs_normalizer.count"].clamp(
    max=NORMALIZER_COUNT
  )


def main(
  input_file: tyro.conf.Positional[str],
  output_file: tyro.conf.Positional[str],
  min_std: float = 0.3,
  learning_rate: float = 3.0e-4,
) -> None:
  """Write the widened checkpoint.

  Args:
    input_file: A Mjlab-Velocity-Flat-LimX-HU-D03 checkpoint.
    output_file: Where to write it, inside a run folder so `train` can resume it.
    min_std: Raise the action noise to at least this, to explore the new task.
    learning_rate: Starting learning rate; the adaptive schedule takes over.
  """
  ckpt = torch.load(input_file, map_location="cpu", weights_only=False)
  _widen(ckpt["actor_state_dict"], FLAT_ACTOR_OBS)
  _widen(ckpt["critic_state_dict"], FLAT_CRITIC_OBS)
  std = ckpt["actor_state_dict"]["distribution.std_param"]
  print(f"action std {std.min():.2f}..{std.max():.2f} -> at least {min_std}")
  std.clamp_(min=min_std)

  # Adam's moments have the old first-layer shape and belong to the old rewards.
  ckpt["optimizer_state_dict"]["state"] = {}
  for group in ckpt["optimizer_state_dict"]["param_groups"]:
    group["lr"] = learning_rate
  # A new training: from step 240,000 the command curriculum would start at its last stage.
  ckpt["iter"] = 0
  ckpt["infos"] = {"env_state": {"common_step_counter": 0}}

  Path(output_file).parent.mkdir(parents=True, exist_ok=True)
  torch.save(ckpt, output_file)
  print(f"wrote {output_file}")


if __name__ == "__main__":
  tyro.cli(main)
```

```python
%%writefile /kaggle/temp/hu_d03_code/hu-d03-rl/scripts/make_terrain_overview.py
"""Tile the terrain clips into one labelled video, green for PASS and red for FAIL.

  python scripts/make_terrain_overview.py videos/terrain_results.json videos/terrains_overview.mp4
"""

import json
import math
from pathlib import Path

import mediapy
import numpy as np
import tyro
from PIL import Image, ImageDraw

import mjlab  # noqa: F401  (Points mediapy at its ffmpeg.)

CELL_WIDTH, CELL_HEIGHT = 320, 180
COLUMNS = 4
FRAME_STRIDE = 2  # Half the frames: keeps the file small.


def _cell(frame: np.ndarray, label: str, passed: bool) -> np.ndarray:
  image = Image.fromarray(frame).resize((CELL_WIDTH, CELL_HEIGHT))
  draw = ImageDraw.Draw(image)
  draw.rectangle((0, 0, CELL_WIDTH, 14), fill=(255, 255, 255))
  draw.text((3, 1), label, fill=(26, 127, 55) if passed else (207, 34, 46))
  return np.asarray(image)


def main(
  results_file: tyro.conf.Positional[str], output_file: tyro.conf.Positional[str]
) -> None:
  results = json.loads(Path(results_file).read_text())
  readers = [mediapy.VideoReader(r["path"]) for r in results]
  for reader in readers:
    reader.__enter__()
  fps = readers[0].metadata.fps / FRAME_STRIDE
  rows = math.ceil(len(results) / COLUMNS)
  blank = np.zeros((CELL_HEIGHT, CELL_WIDTH, 3), dtype=np.uint8)
  labels = [
    f"{r['clip']}  fell {r['fell_percent']:.0f}%  {r['mean_forward_speed']:.1f}/{r['command']} m/s"
    for r in results
  ]
  shape = (rows * CELL_HEIGHT, COLUMNS * CELL_WIDTH)
  with mediapy.VideoWriter(output_file, shape=shape, fps=fps) as writer:
    for i, frames in enumerate(zip(*readers)):
      if i % FRAME_STRIDE:
        continue
      cells = [
        _cell(f, label, r["passed"]) for f, label, r in zip(frames, labels, results)
      ]
      cells += [blank] * (rows * COLUMNS - len(cells))
      writer.add_image(
        np.vstack([
          np.hstack(cells[j : j + COLUMNS]) for j in range(0, len(cells), COLUMNS)
        ])
      )
  for reader in readers:
    reader.__exit__(None, None, None)
  print(f"wrote {output_file}")


if __name__ == "__main__":
  tyro.cli(main)
```

## 6. Train

Logs, checkpoints (every 50 iterations) and ONNX exports go to `/kaggle/working/hu_d03_rl_rough/logs/rsl_rl/hu_d03_velocity_rough/<date_time>/`.
Training is stopped after `TIME_BUDGET_HOURS`; the newest checkpoint is kept.

The cell shows one progress line, updated every 10 s. The percentage is of whichever limit comes first,
`MAX_ITERATIONS` or `TIME_BUDGET_HOURS`. `terrain` is the mean difficulty level, from 0 up to 9; it should climb steadily.
The full rsl_rl output goes to `hu_d03_rl_rough/train.log`

```python
cmd = (f"MUJOCO_GL=egl train {TASK} --env.scene.num-envs {NUM_ENVS} "
       f"--agent.max-iterations {MAX_ITERATIONS} --agent.logger tensorboard")
if USE_TWO_GPUS and NUM_GPUS >= 2:
  cmd += ' --gpu-ids "[0, 1]"'
if RESUME_RUN is not None:
  cmd += f" --agent.resume True --agent.load-run {RESUME_RUN.name}"
else:
  cmd += f" --agent.resume True --agent.load-run {WARM_RUN} --agent.load-checkpoint model_0.pt"
import html, re
from datetime import timedelta
import IPython.display as ipd

class TrainProgress:
  """Parses rsl_rl's per-iteration log and shows one progress line."""
  ANSI = re.compile(r"\x1b\[[0-9;]*m")
  FIELDS = {
    "iteration": re.compile(r"Learning iteration (\d+)/(\d+)"),
    "reward": re.compile(r"Mean reward:\s+(-?[\d.]+)"),
    "length": re.compile(r"Mean episode length:\s+([\d.]+)"),
    "fps": re.compile(r"Steps per second:\s+([\d.]+)"),
    "terrain": re.compile(r"terrain_levels/mean:\s+([\d.]+)"),
  }

  def __init__(self, budget_s):
    self.budget_s = budget_s
    self.values = {}
    self.handle = ipd.display(ipd.HTML("<pre>starting training...</pre>"), display_id=True)

  def on_line(self, line):
    line = self.ANSI.sub("", line)
    for key, pattern in self.FIELDS.items():
      m = pattern.search(line)
      if m:
        self.values[key] = m.groups()

  def on_tick(self, elapsed):
    v = self.values
    if "iteration" not in v:
      text = f"setting up (compiling GPU kernels)...  {timedelta(seconds=int(elapsed))}"
    else:
      it, total = (int(x) for x in v["iteration"])
      frac = min(1.0, max((it + 1) / total, elapsed / self.budget_s))
      bar = "█" * int(frac * 30) + "░" * (30 - int(frac * 30))
      text = f"{bar} {frac * 100:5.1f}%  iter {it + 1}/{total}"
      if "reward" in v:
        text += f"  reward {float(v['reward'][0]):.2f}  ep len {float(v['length'][0]):.0f}/1000"
      if "terrain" in v:
        text += f"  terrain {float(v['terrain'][0]):.2f}"
      if "fps" in v:
        text += f"  {float(v['fps'][0]):,.0f} steps/s"
      text += f"  {timedelta(seconds=int(elapsed))} / {timedelta(seconds=int(self.budget_s))}"
    self.handle.update(ipd.HTML(f"<pre>{html.escape(text)}</pre>"))

progress = TrainProgress(TIME_BUDGET_HOURS * 3600)
sh(cmd, cwd=WORK, max_seconds=TIME_BUDGET_HOURS * 3600, log_file=WORK / "train.log",
   on_line=progress.on_line, on_tick=progress.on_tick)
```

## 8. Results: does it work on every terrain?

Records the newest checkpoint on each terrain, at a fixed difficulty, on a straight-line command: 0.5 is the middle of
the curriculum and 1.0 its hardest level (15 cm stairs, 24° slopes). Each clip simulates `EVAL_ENVS` robots and the camera
follows the first one, not a hand-picked one, so the table describes all of them:

* **Fell:** share of the robots that fell at least once in the clip.
* **Speed:** mean forward speed against the command. A low speed with few falls means the robot is stuck or hesitating.
* **Result:** `PASS` when at most `MAX_FELL_PERCENT` fell and the speed reached `MIN_SPEED_RATIO` of the command.

Robots start on a random tile, so one near the edge of the 40 m area can finish its clip on flat ground; the
hard-terrain numbers are therefore slightly optimistic. This takes about 30–40 minutes. Everything is also in the zip in section 9.

```python
import json, math

EVAL_ENVS = 64
EVAL_SECONDS = 12
MAX_FELL_PERCENT = 10.0
MIN_SPEED_RATIO = 0.7

# (clip, terrain, difficulty, forward speed in m/s)
CLIPS = [("walk_flat", "flat", 0.0, 1.0), ("run_flat", "flat", 0.0, 2.5), ("run_rough", "random_rough", 0.5, 2.0)]
for terrain, speed in [("stairs_up", 0.8), ("stairs_down", 0.8), ("slope_up", 1.0), ("slope_down", 1.0),
                       ("random_rough", 1.0), ("waves", 1.0), ("obstacles", 0.8)]:
  CLIPS += [(f"walk_{terrain}_{level}", terrain, difficulty, speed) for level, difficulty in [("medium", 0.5), ("hard", 1.0)]]

def latest_checkpoint():
  runs = sorted(p for p in LOG_ROOT.glob("*/") if list(p.glob("model_*.pt")))
  ckpts = sorted(runs[-1].glob("model_*.pt"), key=iteration)
  print("checkpoint:", ckpts[-1])
  return ckpts[-1]

def evaluate_terrains():
  checkpoint = latest_checkpoint()
  results = []
  for clip, terrain, difficulty, speed in CLIPS:
    print("---", clip)
    stats_file = WORK / "videos" / "terrains" / f"{clip}.json"
    out = video(f"videos/terrains/{clip}.mp4", f"--checkpoint-file {checkpoint}", f"--terrain {terrain}",
                f"--difficulty {difficulty}", f"--forward-speed {speed}", f"--seconds {EVAL_SECONDS}",
                f"--num-envs {EVAL_ENVS}", f"--stats-file {stats_file}", "--width 854 --height 480", show=False)
    if out is None or not stats_file.exists():
      continue  # Reported as missing in the table.
    stats = json.loads(stats_file.read_text())
    passed = stats["fell_percent"] <= MAX_FELL_PERCENT and stats["mean_forward_speed"] >= MIN_SPEED_RATIO * speed
    results.append({"clip": clip, "path": str(out), "terrain": terrain, "difficulty": difficulty,
                    "command": speed, "passed": passed, **stats})
  (WORK / "videos" / "terrain_results.json").write_text(json.dumps(results, indent=1))
  return results

try:
  RESULTS = evaluate_terrains()
except Exception as e:
  RESULTS = []
  print("Could not evaluate the terrains:", e)
```

## Tuning the gait

Change the weights in the `velocity_env_cfgs.py` cell above and start a new version. To keep the progress so far, add
the previous version as Input: a resume accepts new reward weights.

| What you see | Change in `hu_d03_rough_env_cfg` |
|---|---|
| Arms still stiff | Raise `arm_leg_swing` weight (0.75 → 1.5) |
| Crouched, bent knees | Raise `leg_extension` weight (0.5 → 1.0), or `target_height` (0.90 → 0.905) |
| Short, quick steps | Raise `air_time` weight (0.25 → 0.5) or `threshold_min` (0.1 → 0.2) |
| Limping | Stronger `gait_symmetry` weight (-2 → -5) |
| Falls a lot, terrain level stuck | Lower the gait weights, as they compete with balance; or postpone the speed stages |
| Never runs with a flight phase | Raise `contact_pattern` weight; lower `mechanical_power` |
